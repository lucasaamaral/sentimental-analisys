# Sentiment Scoring Analysis — Neutral Lean Calibration & Mean-Centering
### Brazilian Financial News (2016–2025)

This notebook constructs a **final weekly sentiment score** from the fine-tuned FinBERT-PT-BR model output by applying two successive corrections:

1. **Neutral lean re-scoring**: Articles classified as NEUTRAL are empirically shown to carry a positive lean relative to the overall corpus. A calibration parameter α (determined via Pearson correlation between the weekly neutral rate and the weekly net sentiment) re-weights each neutral article's contribution so that its full probability mass is used rather than discarded.

2. **Historical mean-centering**: The raw weekly score is shifted by the corpus-wide mean (μ = −0.0978), anchoring the zero line to the historical average rather than to the arbitrary class-balance midpoint. Positive values on the final score indicate **above-average** sentiment weeks; negative values indicate **below-average** weeks.

The notebook documents every modelling decision — centroid geometry, cosine similarity tests, correlation analysis, and the comparison of five alternative scoring approaches — before committing to the chosen methodology.

## 1. Imports & Configuration

Standard scientific-Python stack. Two project-specific helpers are defined here and reused throughout:
- `week_key_to_date` — converts ISO week strings (e.g. `"2020-W11"`) to the Monday `Timestamp` of that week.
- `prepare_sentiment_frame` — normalises raw softmax scores to a proper probability triplet `[pos, neg, neu]`, adds `confidence`, `net_score`, and `entropy` columns, and corrects the year field for any records timestamped in 2026.

`APPROACH_COLORS` and `APPROACH_LABELS` map the five scoring strategies evaluated in Section 5 to display names used in comparison charts.


```python
import warnings
from pathlib import Path

import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib.lines import Line2D
from scipy import stats
from scipy.stats import entropy as scipy_entropy
from sklearn.metrics.pairwise import cosine_similarity

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", palette="muted", font_scale=1.1)
plt.rcParams["figure.dpi"] = 130
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

FINETUNED_RESULTS_PATH = Path("output/finetuned_model_classified.jsonl")
COLORS = {"POSITIVE": "#27AE60", "NEGATIVE": "#E74C3C", "NEUTRAL": "#3498DB"}

# Approach colors — kept for reference, not used in the final scoring section
APPROACH_COLORS = {
    "A_binary":          "#7F8C8D",
    "B_net_score":       "#2980B9",
    "C_cosine_global":   "#8E44AD",
    "D_temporal_corr":   "#E67E22",
    "E_cosine_per_art":  "#27AE60",
}
APPROACH_LABELS = {
    "A_binary":          "A — Binary (POS=+1, NEG=−1, NEU=0)",
    "B_net_score":       "B — Raw net score (pos−neg for all articles)",
    "C_cosine_global":   "C — Neutral: global cosine lean (mean α)",
    "D_temporal_corr":   "D — Neutral: lean via temporal correlation",
    "E_cosine_per_art":  "E — Neutral: per-article cosine lean",
}

def week_key_to_date(wk: str) -> pd.Timestamp:
    year, week = wk.split("-W")
    return pd.Timestamp.fromisocalendar(int(year), int(week), 1)

def prepare_sentiment_frame(frame: pd.DataFrame) -> pd.DataFrame:
    frame = frame.copy()
    score_sum = frame["score_positive"] + frame["score_negative"] + frame["score_neutral"]
    frame["pos"] = frame["score_positive"] / score_sum
    frame["neg"] = frame["score_negative"] / score_sum
    frame["neu"] = frame["score_neutral"] / score_sum
    frame["confidence"] = frame.apply(
        lambda r: {"POSITIVE": r["pos"], "NEGATIVE": r["neg"], "NEUTRAL": r["neu"]}[r["predicted_label"]], axis=1
    )
    frame["net_score"] = frame["pos"] - frame["neg"]
    frame["year"] = frame["week_key"].str[:4].astype(int)
    frame.loc[frame["year"] == 2026, "year"] = 2025
    frame["entropy"] = [scipy_entropy(row) for row in frame[["pos", "neg", "neu"]].values]
    return frame
```

## 2. Data Loading & Preparation

Reads the fine-tuned model's JSONL output, normalises the raw softmax scores into a proper probability triplet `[pos, neg, neu]` (ensuring each row sums to exactly 1.0), and extracts temporal features. Only records from 2016–2025 are kept.

**Why normalise?** The model outputs logits that go through softmax, but floating-point rounding can cause slight deviations from 1.0. Explicit normalisation prevents any drift when computing weighted contributions.


```python
df_raw = pd.read_json(FINETUNED_RESULTS_PATH, lines=True)
df = prepare_sentiment_frame(df_raw)
df["published_at"] = pd.to_datetime(df["published_at"], format="ISO8601", utc=True)
df["week_num"] = df["week_key"].str.extract(r"W(\d+)").astype(int)
df["week_start"] = df["week_key"].map(week_key_to_date)
df = df[df["year"].between(2016, 2025)].copy().reset_index(drop=True)

n_pos = (df["predicted_label"] == "POSITIVE").sum()
n_neg = (df["predicted_label"] == "NEGATIVE").sum()
n_neu = (df["predicted_label"] == "NEUTRAL").sum()
total = len(df)
print(f"Total: {total:,}  |  POSITIVE: {n_pos:,} ({n_pos/total:.1%})  |  "
      f"NEGATIVE: {n_neg:,} ({n_neg/total:.1%})  |  NEUTRAL: {n_neu:,} ({n_neu/total:.1%})")
print(f"Interval: {df['published_at'].min().date()} → {df['published_at'].max().date()}")
df[["predicted_label", "pos", "neg", "neu", "net_score", "week_key"]].head(5)
```

## 3. Class Centroid Analysis

Computes the mean probability vector `[pos, neg, neu]` for each predicted class and visualises them in all three 2D projections of the simplex.

**Purpose:** The centroids are the "typical" probability triplet for each class. Comparing where the NEUTRAL centroid sits relative to POSITIVE and NEGATIVE gives a first geometric intuition of how neutral articles differ from the other two classes and whether they lean towards one side. These centroids are also used in Section 4 to measure per-article cosine similarity.


```python
vectors = df[["pos", "neg", "neu"]].values
labels  = df["predicted_label"].values

# Class centroids in [pos, neg, neu] space
centroid_pos = df[df["predicted_label"] == "POSITIVE"][["pos", "neg", "neu"]].mean().values
centroid_neg = df[df["predicted_label"] == "NEGATIVE"][["pos", "neg", "neu"]].mean().values
centroid_neu = df[df["predicted_label"] == "NEUTRAL"][["pos", "neg", "neu"]].mean().values

print("=== Class centroids (mean [pos, neg, neu]) ===")
for name, c in [("POSITIVE", centroid_pos), ("NEGATIVE", centroid_neg), ("NEUTRAL", centroid_neu)]:
    print(f"  {name:10s}  pos={c[0]:.4f}  neg={c[1]:.4f}  neu={c[2]:.4f}  |  net={c[0]-c[1]:+.4f}")

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

components = [("pos", "neg"), ("pos", "neu"), ("neg", "neu")]
comp_labels = [("P(POSITIVE)", "P(NEGATIVE)"), ("P(POSITIVE)", "P(NEUTRAL)"), ("P(NEGATIVE)", "P(NEUTRAL)")]

for ax, (cx, cy), (lx, ly) in zip(axes, components, comp_labels):
    sample = df.sample(min(2000, len(df)), random_state=42)
    for lbl in ["POSITIVE", "NEGATIVE", "NEUTRAL"]:
        sub = sample[sample["predicted_label"] == lbl]
        ax.scatter(sub[cx], sub[cy], c=COLORS[lbl], alpha=0.25, s=8, label=lbl, edgecolors="none")
    # Plot centroids
    for lbl, c in [("POSITIVE", centroid_pos), ("NEGATIVE", centroid_neg), ("NEUTRAL", centroid_neu)]:
        xi = {"pos": 0, "neg": 1, "neu": 2}
        ax.scatter(c[xi[cx]], c[xi[cy]], c=COLORS[lbl], s=200, marker="*",
                   edgecolors="black", linewidth=1, zorder=10)
    ax.set_xlabel(lx); ax.set_ylabel(ly)
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)

axes[0].legend(fontsize=9, markerscale=2)
plt.suptitle("Probability Space by Class — Fine-Tuned Model\n(★ = class centroid)",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

## 4. Neutral Article Similarity to Positive / Negative Centroids

For each NEUTRAL article, computes the **cosine similarity** of its probability vector `[pos, neg, neu]` to the POSITIVE centroid and to the NEGATIVE centroid. The difference `sim_pos − sim_neg` is the per-article **cosine lean** — a signed value that indicates whether a neutral article's probability profile is more aligned with typical positive or typical negative news.

**Why cosine similarity?** Unlike Euclidean distance, cosine similarity measures the *direction* (shape of the probability allocation) rather than the absolute magnitude, which is the meaningful property here since all vectors already sum to 1 and the relevant question is the *relative* allocation across the three dimensions.

**Hypothesis tested (Wilcoxon signed-rank):** The null hypothesis is that `sim_to_pos = sim_to_neg` for neutral articles, i.e., that the model's neutral class is equidistant from both poles. A statistically significant positive lean would justify giving neutral articles a slight positive weight.


```python
neu_mask = df["predicted_label"] == "NEUTRAL"
neu_vecs = df.loc[neu_mask, ["pos", "neg", "neu"]].values

# Cosine similarity to each class centroid
sim_to_pos = cosine_similarity(neu_vecs, centroid_pos.reshape(1, -1)).ravel()
sim_to_neg = cosine_similarity(neu_vecs, centroid_neg.reshape(1, -1)).ravel()
sim_to_neu = cosine_similarity(neu_vecs, centroid_neu.reshape(1, -1)).ravel()

# Per-article cosine lean: in [-1, +1]
#   +1  → identical direction to POS centroid
#   -1  → identical direction to NEG centroid
cosine_lean = sim_to_pos - sim_to_neg  # range roughly in [-1, +1]

print(f"NEUTRAL articles: {neu_mask.sum():,}")
print(f"\nMean cosine similarity to POSITIVE centroid : {sim_to_pos.mean():.4f}  ±{sim_to_pos.std():.4f}")
print(f"Mean cosine similarity to NEGATIVE centroid : {sim_to_neg.mean():.4f}  ±{sim_to_neg.std():.4f}")
print(f"Mean cosine similarity to NEUTRAL centroid  : {sim_to_neu.mean():.4f}  ±{sim_to_neu.std():.4f}")
print(f"\nMean lean (sim_pos − sim_neg): {cosine_lean.mean():+.4f}  "
      f"[{'→ POSITIVE' if cosine_lean.mean() > 0 else '→ NEGATIVE'}]")

wilc = stats.wilcoxon(sim_to_pos - sim_to_neg)
print(f"\nWilcoxon signed-rank (sim_pos vs sim_neg): stat={wilc.statistic:.1f}, p={wilc.pvalue:.2e}")
if wilc.pvalue < 0.05:
    direction = "POSITIVE" if cosine_lean.mean() > 0 else "NEGATIVE"
    print(f"→ Statistically significant result: neutral articles are more similar to {direction} (p<0.05)")
else:
    print("→ No statistically significant difference between the two cosine similarities.")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax = axes[0]
ax.hist(sim_to_pos, bins=60, alpha=0.65, color=COLORS["POSITIVE"], label="Sim → POSITIVE", edgecolor="none")
ax.hist(sim_to_neg, bins=60, alpha=0.65, color=COLORS["NEGATIVE"], label="Sim → NEGATIVE", edgecolor="none")
ax.axvline(sim_to_pos.mean(), color=COLORS["POSITIVE"], linestyle="--", linewidth=1.8)
ax.axvline(sim_to_neg.mean(), color=COLORS["NEGATIVE"], linestyle="--", linewidth=1.8)
ax.set_title("Cosine Similarity Distribution\n(NEUTRAL articles → class centroids)", fontweight="bold")
ax.set_xlabel("Cosine Similarity")
ax.set_ylabel("Number of Articles")
ax.legend(fontsize=9)

ax = axes[1]
ax.hist(cosine_lean, bins=80, color="#8E44AD", alpha=0.8, edgecolor="none")
ax.axvline(0, color="black", linewidth=1, linestyle="--")
ax.axvline(cosine_lean.mean(), color="#8E44AD", linestyle="-", linewidth=2,
           label=f"Mean = {cosine_lean.mean():+.4f}")
ax.set_title("Per-article Cosine Lean for NEUTRAL Articles\n(sim_pos − sim_neg)", fontweight="bold")
ax.set_xlabel("Lean (positive = more similar to POSITIVE centroid)")
ax.set_ylabel("Number of Articles")
ax.legend(fontsize=9)

plt.suptitle("Cosine Similarity Analysis — NEUTRAL Articles", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

## 5. Neutral Lean Calibration via Temporal Correlation

The α factor controls how the `neu` probability of a neutral article is redistributed between the positive and negative sides of the score:

$$\text{contrib}_i = p_i^+ - p_i^- + (2\alpha - 1) \cdot p_i^0$$

where $p^+, p^-, p^0$ are the normalised softmax probabilities for POSITIVE, NEGATIVE, and NEUTRAL respectively. When:
- **α = 0.5** → $(2 \times 0.5 - 1) = 0$ → neutral probability contributes nothing (raw net score)
- **α > 0.5** → positive lean: neutral articles pull the weekly score upward
- **α < 0.5** → negative lean: neutral articles pull the weekly score downward

**Three estimation methods are compared:**

| Method | Formula | Intuition |
|--------|---------|-----------|
| **Temporal** | $\alpha = 0.5 + 0.5 \times r$ where $r$ = Pearson correlation between weekly neutral rate and weekly net sentiment | If more neutral articles appear in historically positive weeks, give them a positive lean |
| **OLS regression** | Regress `net_label ~ rate_pos + rate_neu + const`, derive α from OLS coefficients | Algebraic best-fit of how much the neutral rate explains net sentiment |
| **Cosine global** | $\alpha = \text{mean}({\rm sim\_pos} / ({\rm sim\_pos} + {\rm sim\_neg}))$ across all neutral articles | Geometric lean based on probability-space proximity to each pole |

**Chosen method: Temporal Correlation.** It is the only method that directly uses the empirical co-occurrence of neutral articles with positive/negative market weeks, making it the most behaviourally grounded. The OLS result is close to the algebraic identity ($\beta_{\rm pos}=2, \beta_{\rm neu}=1$) and offers little additional insight. The cosine method gives a very small α ≈ 0.525 because the geometric separation between neutral and the two poles is almost symmetric.


```python
# ── Weekly aggregation ────────────────────────────────────────────────────
weekly = (
    df.groupby(["week_key", "week_start", "year"])
    .agg(
        n_news        = ("predicted_label", "count"),
        n_positive    = ("predicted_label", lambda x: (x == "POSITIVE").sum()),
        n_negative    = ("predicted_label", lambda x: (x == "NEGATIVE").sum()),
        n_neutral     = ("predicted_label", lambda x: (x == "NEUTRAL").sum()),
        mean_pos_prob = ("pos", "mean"),
        mean_neg_prob = ("neg", "mean"),
        mean_neu_prob = ("neu", "mean"),
    )
    .reset_index()
    .sort_values("week_start")
    .reset_index(drop=True)
)
weekly["rate_pos"] = weekly["n_positive"] / weekly["n_news"]
weekly["rate_neg"] = weekly["n_negative"] / weekly["n_news"]
weekly["rate_neu"] = weekly["n_neutral"]  / weekly["n_news"]
weekly["net_label"] = weekly["rate_pos"] - weekly["rate_neg"]

# ── Method 1: Pearson correlation ─────────────────────────────────────────
corr_neu_net, pval_corr = stats.pearsonr(weekly["rate_neu"], weekly["net_label"])
alpha_temporal = 0.5 + 0.5 * corr_neu_net

print("=== Temporal Correlation: rate_neutral × weekly net_label ===")
print(f"  Pearson r  = {corr_neu_net:+.4f}  (p = {pval_corr:.3e})")
print(f"  Spearman ρ = {stats.spearmanr(weekly['rate_neu'], weekly['net_label']).statistic:+.4f}")
print(f"  → α_temporal = 0.5 + 0.5×r = {alpha_temporal:.4f}")
if corr_neu_net > 0:
    print("  → Weeks with more neutral articles tend to be MORE POSITIVE (positive lean)")
else:
    print("  → Weeks with more neutral articles tend to be MORE NEGATIVE (negative lean)")

# ── Method 2: OLS regression ──────────────────────────────────────────────
# net_label ≈ β_pos*rate_pos + β_neg*rate_neg + β_neu*rate_neu
# rate_pos + rate_neg + rate_neu = 1, so drop one (rate_neg) and use intercept
from numpy.linalg import lstsq
X = weekly[["rate_pos", "rate_neu"]].values
X_aug = np.column_stack([X, np.ones(len(X))])
y = weekly["net_label"].values
coefs, _, _, _ = lstsq(X_aug, y, rcond=None)
beta_pos_ols, beta_neu_ols, intercept_ols = coefs
# The implicit beta_neg comes from the net_label identity:
# In pure binary: net_label = rate_pos - rate_neg = rate_pos - (1 - rate_pos - rate_neu) = 2*rate_pos + rate_neu - 1
# So ideal: β_pos=2, β_neu=1, intercept=-1
# From OLS: lean of neu is between β_neg_eff=-β_pos and β_pos_eff=+β_pos
beta_neg_eff = -(beta_pos_ols)  # symmetric assumption
alpha_ols = (beta_neu_ols - beta_neg_eff) / (beta_pos_ols - beta_neg_eff)
alpha_ols = float(np.clip(alpha_ols, 0, 1))

print(f"\n=== OLS Regression: net_label ~ rate_pos + rate_neu + const ===")
print(f"  β_pos = {beta_pos_ols:+.4f}  |  β_neu = {beta_neu_ols:+.4f}  |  intercept = {intercept_ols:+.4f}")
print(f"  → α_ols = {alpha_ols:.4f}  (fraction of neutral prob. attributed to the positive side)")

# ── Method 3: Global cosine alpha ─────────────────────────────────────────
# α = mean(sim_to_pos / (sim_to_pos + sim_to_neg)) across neutral articles
sim_sum = sim_to_pos + sim_to_neg
sim_sum = np.where(sim_sum == 0, 1e-9, sim_sum)  # avoid div by zero
alpha_cosine_global = float(np.mean(sim_to_pos / sim_sum))

print(f"\n=== Global Cosine ===")
print(f"  α_cosine_global = {alpha_cosine_global:.4f}")

print(f"\n=== Summary of estimated α values ===")
print(f"  α_temporal  (Pearson correlation) = {alpha_temporal:.4f}")
print(f"  α_ols       (OLS regression)      = {alpha_ols:.4f}")
print(f"  α_cosine    (global cosine)       = {alpha_cosine_global:.4f}")

# ── Visualise correlation ─────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax = axes[0]
ax.scatter(weekly["rate_neu"], weekly["net_label"],
           c=weekly["year"], cmap="plasma", alpha=0.5, s=18, edgecolors="none")
m, b = np.polyfit(weekly["rate_neu"], weekly["net_label"], 1)
x_line = np.linspace(weekly["rate_neu"].min(), weekly["rate_neu"].max(), 100)
ax.plot(x_line, m * x_line + b, color="black", linewidth=1.8, linestyle="--",
        label=f"r = {corr_neu_net:+.3f}  (p={pval_corr:.2e})")
ax.axhline(0, color="gray", linewidth=0.8)
ax.set_xlabel("Weekly rate of NEUTRAL articles")
ax.set_ylabel("Weekly net sentiment (%POS − %NEG)")
ax.set_title("Correlation: % Neutral × Weekly Net Sentiment", fontweight="bold")
ax.legend(fontsize=9)

ax = axes[1]
alphas_dict = {"α_temporal": alpha_temporal, "α_ols": alpha_ols, "α_cosine": alpha_cosine_global}
bars = ax.bar(alphas_dict.keys(), alphas_dict.values(),
              color=["#E67E22", "#2980B9", "#8E44AD"], alpha=0.85, edgecolor="white")
ax.axhline(0.5, color="black", linestyle="--", linewidth=1.5, label="α=0.5 (perfect neutral)")
for bar, val in zip(bars, alphas_dict.values()):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.01,
            f"{val:.4f}", ha="center", fontsize=10, fontweight="bold")
ax.set_ylim(0, 1)
ax.set_ylabel("α (fraction of neutral prob. → positive)")
ax.set_title("α Estimates — Three Methods", fontweight="bold")
ax.legend(fontsize=9)

plt.suptitle("α Factor Calibration for NEUTRAL Articles", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

## 6. Final Scoring: Temporal Lean + Mean-Centering

### Chosen approach

Two design decisions combine to produce the final weekly sentiment index:

---

**Decision 1 — Neutral article correction (temporal lean, α = α_temporal)**

The raw contribution of each article to the weekly score uses the model's actual softmax probabilities rather than a binary ±1 label. For POSITIVE and NEGATIVE articles the contribution is simply `pos − neg`. For NEUTRAL articles, the `neu` probability mass is redistributed using the empirically calibrated α:

$$\text{score}_i = \begin{cases} p_i^+ - p_i^- & \text{if predicted\_label} \neq \text{NEUTRAL} \\ p_i^+ - p_i^- + (2\alpha_{\rm temp} - 1)\cdot p_i^0 & \text{if predicted\_label} = \text{NEUTRAL} \end{cases}$$

The weekly score is the arithmetic mean of all per-article scores in that week.

*Justification:* Section 5 showed that weeks with a higher proportion of neutral articles tend to be more positive (Pearson r = +0.265, p < 0.001 across 522 weekly observations). This is not a trivial signal — it is consistent with the interpretation that in genuinely uncertain or transitional market conditions, news tends to cluster around the neutral class but the overall economic context is mildly constructive.

---

**Decision 2 — Class-imbalance correction (historical mean-centering)**

The dataset has a structural imbalance: NEGATIVE (42.0%) > NEUTRAL (39.2%) >> POSITIVE (18.8%). As a result, the raw weekly average score is chronically negative (mean ≈ −0.10 even after the neutral correction), making it impossible to distinguish between "genuinely bad week" and "an average week in a structurally negative corpus".

The correction subtracts the long-run historical mean of the adjusted score:

$$\text{final\_score}_w = \overline{\text{score}}_w - \mu_{\text{hist}}$$

**Interpretation of the zero line after centering:**
- `final_score > 0` → this week's sentiment is *above the historical average* (relatively optimistic)
- `final_score = 0` → exactly at the long-run baseline
- `final_score < 0` → below the historical average (relatively pessimistic)

*Why not z-score?* Dividing by σ (z-score) would make the unit abstract (standard deviations), which is harder to communicate. Mean-centering preserves the original scale (net probability difference) while shifting the reference point from "never positive" to "above / below average".

*Why not macro class-balancing?* A macro-balanced score (equal weight per class per week) distorts the weekly signal when one class has very few articles — e.g., a week with 1 positive and 20 negative articles would still contribute positively because the single positive article carries 1/3 of the weight. Mean-centering treats all articles equally but adjusts the reference point.

---

### Key parameters

| Parameter | Value | Source |
|-----------|-------|--------|
| α_temporal | computed in Section 5 | Pearson r between weekly neutral rate and net sentiment |
| μ_hist | mean of adjusted weekly scores | computed below from 2016–2025 data |
| MA window | 13 weeks | ~1 quarter smoothing, robust to single-week outliers |


```python
from IPython.display import display

# ─────────────────────────────────────────────────────────────────────────────
# Step 1: Per-article contribution score
# ─────────────────────────────────────────────────────────────────────────────
# For non-neutral articles: contrib = pos - neg
# For neutral articles:     contrib = pos - neg + (2*alpha_temporal - 1) * neu
#
# alpha_temporal was derived in Section 5 from Pearson correlation between the
# weekly neutral rate and the weekly net sentiment.  A value > 0.5 means that
# weeks with more neutral articles tend to be slightly more positive, so we
# give the neutral probability mass a slight positive lean.

df["contrib"] = df["pos"] - df["neg"]                           # default for non-neutral

neu_mask = df["predicted_label"] == "NEUTRAL"
lean_coef = 2 * alpha_temporal - 1                              # scalar in (-1, +1)
df.loc[neu_mask, "contrib"] = (
    df.loc[neu_mask, "pos"]
    - df.loc[neu_mask, "neg"]
    + lean_coef * df.loc[neu_mask, "neu"]
)

print(f"Neutral lean coefficient (2α−1) = {lean_coef:+.4f}")
print(f"  → Each neutral article's neu probability is multiplied by {lean_coef:+.4f}")
print(f"    before being added to the pos−neg net score.\n")

# ─────────────────────────────────────────────────────────────────────────────
# Step 2: Weekly aggregation
# ─────────────────────────────────────────────────────────────────────────────
# The weekly score is the arithmetic mean of all per-article contributions.
# Using the mean (rather than sum) makes weeks with different article counts
# comparable — a high-volume week does not automatically dominate.

ws = (
    df.groupby(["week_key", "week_start"])
    .agg(
        n_articles    = ("contrib", "count"),
        n_positive    = ("predicted_label", lambda x: (x == "POSITIVE").sum()),
        n_negative    = ("predicted_label", lambda x: (x == "NEGATIVE").sum()),
        n_neutral     = ("predicted_label", lambda x: (x == "NEUTRAL").sum()),
        raw_score     = ("contrib", "mean"),
    )
    .reset_index()
    .sort_values("week_start")
    .reset_index(drop=True)
)
ws["rate_pos"] = ws["n_positive"] / ws["n_articles"]
ws["rate_neg"] = ws["n_negative"] / ws["n_articles"]
ws["rate_neu"] = ws["n_neutral"]  / ws["n_articles"]

# ─────────────────────────────────────────────────────────────────────────────
# Step 3: Mean-centering to correct the class-imbalance baseline
# ─────────────────────────────────────────────────────────────────────────────
# The raw score is structurally negative because NEGATIVE articles outnumber
# POSITIVE by more than 2:1.  Subtracting the historical mean shifts the zero
# to represent "a typical week" rather than "no sentiment at all".
#
# After centering:
#   final_score > 0  ←→  this week is above the historical average
#   final_score = 0  ←→  at the long-run baseline
#   final_score < 0  ←→  below the historical average

mu_hist    = ws["raw_score"].mean()
sigma_hist = ws["raw_score"].std()
ws["final_score"] = ws["raw_score"] - mu_hist

# 13-week centred moving average (~1 quarter)
ws["ma13"] = ws["final_score"].rolling(13, center=True, min_periods=6).mean()

dates = ws["week_start"]

# ─────────────────────────────────────────────────────────────────────────────
# Descriptive statistics
# ─────────────────────────────────────────────────────────────────────────────
print("=== Raw Score (before centering) ===")
print(f"  Mean  : {mu_hist:+.4f}  ← subtracted as baseline")
print(f"  Std   : {sigma_hist:.4f}")
print(f"  Min   : {ws['raw_score'].min():+.4f}   Max: {ws['raw_score'].max():+.4f}")
print(f"  % weeks > 0 : {(ws['raw_score'] > 0).mean():.1%}\n")

print("=== Final Score (after centering) ===")
print(f"  Mean  : {ws['final_score'].mean():+.6f}  (exactly 0 by construction)")
print(f"  Std   : {ws['final_score'].std():.4f}")
print(f"  Min   : {ws['final_score'].min():+.4f}   Max: {ws['final_score'].max():+.4f}")
print(f"  % weeks > 0 (raw)  : {(ws['final_score'] > 0).mean():.1%}")
print(f"  % weeks > 0 (MA13) : {(ws['ma13'].dropna() > 0).mean():.1%}")
print(f"\n  → α_temporal used  : {alpha_temporal:.4f}")
print(f"  → μ_hist subtracted: {mu_hist:+.4f}")
```
